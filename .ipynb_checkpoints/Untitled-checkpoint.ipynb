```python
shopping_list = ["apple", "banana", "apple", "orange", "banana", "grape"]

unique_shopping_list = list(set(shopping_list))

print(unique_shopping_list)
```

```python
course_a = {"John", "Alice", "Bob", "David"}
course_b = {"Alice", "Eve", "Charlie", "David"}

only_course_a = course_a.difference(course_b)

only_one_course = course_a.symmetric_difference(course_b)

print("Students in Course A but not in Course B:", only_course_a)
print("Students only in one of the two courses:", only_one_course)
```

```python
numbers = (10, 20, 30, 40, 50, 60, 70)

first_three = numbers[:3]
print("First 3 elements:", first_three)

last_three = numbers[-3:]
print("Last 3 elements:", last_three)

alternate = numbers[::2]
print("Alternate elements:", alternate)
```

```python
A = {"Ali", "Sara"}
B = {"Ahmed", "Zara"}

combined_set = A.union(B)

print("Combined set:", combined_set)
```

```python
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7}

common_elements = A.intersection(B)
print("Common elements:", common_elements)

only_in_A = A.difference(B)
print("Elements only in Set A:", only_in_A)

only_in_B = B.difference(A)
print("Elements only in Set B:", only_in_B)
```
